```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def vandercor(n):
    res = []
    for i in range(1,n+1):
        num = 0
        for j in range(32):
            if(i&2**j):
                num+=1/(2**(j+1))

        res.append(num)

    return res

def generalradical(n,b):
    res = []
    for i in range(1,n+1):
        num = 0
        val = i
        f = 1/b
        while(val > 0):
            num += (val%b)*f
            f /= b
            val //= b

        res.append(num)

    return res

        
```

```python
res = vandercor(25)
print(res)
```

```python

res2 = vandercor(1000)
plt.scatter(res2[:-1],res2[1:])
plt.xlabel("xi")
plt.ylabel("xi+1")
plt.grid()
plt.show()
```

```python
res3 = vandercor(100)
res4 = vandercor(100000)


def lcg(a,c,m,n):
    res = []
    x0 = 42
    for i in range(n):
        x0 = (a*x0+c)%m
        res.append(x0/m)

    return res


fig,axes = plt.subplots(2,2,figsize = (12,8))

axes[0,0].set_title("van der corrupt(100)")
axes[0,0].hist(res3,bins = 50,rwidth = 0.85)

axes[0,1].set_title("van der corrupt(100000)")
axes[0,1].hist(res4,bins = 50,rwidth = 0.85)

axes[1,0].set_title("lcg generated (100)")
axes[1,0].hist(lcg(1664525,1013904223,2**32,100),bins = 50,rwidth = 0.85)

axes[1,1].set_title("lcg generated (100000)")
axes[1,1].hist(lcg(1664525,1013904223,2**32,1000),bins = 50,rwidth = 0.85)

plt.show()

```
