```python
import numpy as np
import matplotlib.pyplot as mp
import pandas as pd
```

```python
def func1(x0,a,b,m) : 
    res = []
    curr = x0
    res.append(x0)
    for i in range (0,20) :
        res.append(curr)
        curr =  ( a*curr + b ) % m

    return res

```

```python
for x in range(0,10):
    res = func1(x,6,0,11)

    print(f"Result for x0 = {x} , a = 6 , b = 6 , m = 11 : {res} no of unique numbers are {len(set(res))}")

for x in range(0,10):
    res = func1(x,3,0,11)
    print(f"Result for x0 = {x}, a = 3, b = 0, m = 11: {res} no of unique numbers are {len(set(res))}")
```
