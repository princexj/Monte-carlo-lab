```python
import numpy as np
```

```python
#part 1a)
def func1(m):
    np.random.seed(40)
    res = np.exp(np.random.uniform(0,1,m))
    resmean = np.mean(res)
    resvar = np.var(res,ddof = 1)/m
    # print(f"for m = {m} sampling variance is = {resvar}")
    return resmean,resvar
```

```python
lst = [10,100,1000,10000,100000]
for i in range(len(lst)):
    func1(lst[i])
```

```python
#part 1b)
def func2(m):
    np.random.seed(40)
    res2 = np.random.uniform(0,1,m)
    res = 0.5*(np.exp(res2)+np.exp(1-res2))
    resmean = np.mean(res)
    resvar = np.var(res,ddof = 1)/m
    # print(f"for m = {m} sampling variance is = {resvar}")
    return resmean,resvar
```

```python
lst = [10,100,1000,10000,100000]
for i in range(len(lst)):
    func2(lst[i]//2)
```

```python
#part 1c)
def func3(m):
    np.random.seed(40)
    u = np.random.uniform(0,1,m)
    yi = np.exp(u)
    umean = np.mean(u)
    umean = 0.5

    cbest = np.cov(u,yi,ddof = 1)[0,1]/np.var(u,ddof = 1)
    res = yi - cbest*(u - umean)
    resmean = np.mean(res)
    resvar = np.var(res,ddof = 1)/m
    # print(f"for m = {m} sampling variance is = {resvar}")
    return resmean,resvar
```

```python
lst = [10,100,1000,10000,100000]
for i in range(len(lst)):
    func3(lst[i])
```

```python
#1d
import pandas as pd

ans = []
for m in lst:
    mean1,var1 = func1(m)
    mean2,var2 = func2(m//2)
    mean3,var3 = func3(m)

    reduced2 = (var1-var2)/var1 * 100
    reduced3 = (var1-var3)/var1 * 100

    ans.append({
        "M": m,
        "Crude Estimate": mean1,
        "Antithetic Estimate": mean2,
        "Control Estimate": mean3,
        "Crude Variance": var1,
        "Antithetic Variance": var2,
        "Control Variance": var3,
        "Antithetic Reduction (%)": reduced2,
        "Control Reduction (%)": reduced3
    })

df = pd.DataFrame(ans)
print(df.to_string())

```
