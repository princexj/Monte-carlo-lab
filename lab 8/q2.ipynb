```python
import numpy as np
from scipy.stats import norm
M = [10,100,10**3,10**4,10**5]
np.random.seed(42)
```

```python
def func1(m):
    ys = np.random.exponential(scale = 1,size = m)
    xs = np.random.normal(loc = ys,scale = 2)

    xs = (xs > 1).astype(int)
    # print(f"for m = {m}the probability p(x > 1) is {sum(xs)/m}")
    samplevar = np.var(xs,ddof = 1)/m
    return sum(xs/m),samplevar


for m in M:
    func1(m)
```

```python
def func2(m):
    ys = np.random.exponential(scale = 1,size = m)
    result = norm.cdf((ys-1)/2)
    # print(f"for m = {m} The probability P(X > 1) is {np.mean(result)}")
    samplevar = np.var(result,ddof = 1)/m
    return np.mean(result),samplevar


for m in M:
    func2(m)
```

```python
#using antithetic variable
def func3(m):
    m = m//2
    us = np.random.uniform(size = m)
    ys1 = -np.log(us)
    ys2 = -np.log(1-us)

    result = 0.5*(norm.cdf((ys1-1)/2) + norm.cdf((ys2-1)/2))
    # print(f"for m = {m} The probability P(X > 1) is {np.mean(result)}")

    samplevar = np.var(result,ddof = 1)/m
    return np.mean(result),samplevar

for m in M:
    func3(m)
    
```

```python
def func4(m):
    # us = np.random.uniform(size = m)
    ys = np.random.exponential(scale = 1,size = m)

    g = norm.cdf((ys-1)/2)
    cstar = np.cov(g,ys,ddof = 1)[0,1]/np.var(ys,ddof = 1)
    result = g - cstar*(ys-1)

    # print(f"for m = {m} The probability P(X > 1) is {np.mean(result)}")
    samplevar = np.var(result,ddof = 1)/m
    return np.mean(result),samplevar

for m in M:
    func4(m)
```

```python
import pandas as pd

ans = []
for m in M: 
    p1,v1 = func1(m)
    p2,v2 = func2(m)
    p3,v3 = func3(m)
    p4,v4 = func4(m)

    ans.append({
        "M": m,
        "Basic Estimate": p1,
        "Basic Variance": v1,
        "Condition Estimate": p2,
        "Condition Variance": v2,
        "Condition Reduction (%)": 100*(1-v2/v1),
        "Antithetic Estimate": p3,
        "Antithetic Variance": v3,
        "Antithetic Reduction (%)": 100*(1-v3/v1),
        "Control Estimate": p4,
        "Control Variance": v4,
        "Control Reduction (%)": 100*(1-v4/v1)
        
    })

ans = pd.DataFrame(ans)
print(ans.to_string(index = False))
```
