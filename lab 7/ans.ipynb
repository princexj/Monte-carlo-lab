```python
import numpy as np
from scipy.integrate import quad
```

```python
def func2(u):
    u = np.exp(np.sqrt(u))
    return u

i,err = quad(func2,0,1)
print(f"the theoritical expected value : {i}")
```

```python
def func1(m):
    print(f"answer for m = ")
    np.random.seed(42)
    res = []
    temp = np.random.uniform(0,1,m)
    res = np.exp(np.sqrt(temp))

    print(f"the mean is : {np.mean(res)}")
    meanv = float(np.mean(res))
    
    s = float(np.std(res,ddof = 1))
    serr = s/np.sqrt(m)
    print(f"the interval  is {( meanv-1.96*serr,meanv+1.96*serr )}")

    print()
```

```python
func1(100)
func1(1000)
func1(10000)
func1(100000)
```

```python
def func2(m):
    m = int(m)
    print(f"answer for m = ")
    np.random.seed(42)
    temp = np.random.uniform(0, 1, m)
    res = 0.5 * (np.exp(np.sqrt(temp)) + np.exp(np.sqrt(1 - temp)))

    print(f"the mean is : {np.mean(res)}")
    meanv = float(np.mean(res))

    s = float(np.std(res, ddof=1))
    serr = s / np.sqrt(m)
    print(f"the interval  is {(( meanv-1.96*serr, meanv+1.96*serr ))}")

    print()
```

```python
func2(10**2 / 2)
func2(10**2)
func2(10**3 / 2)
func2(10**3)
func2(10**4 / 2)
func2(10**4)
func2(10**5 / 2)
func2(10**5)
```
