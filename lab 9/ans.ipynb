```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# question 1
np.random.seed(42)
n = 5000
ans = np.zeros((10,5001))
dt = 5/5000
t = np.linspace(0,5,5001)

for i in range(1,5001):
    z = np.random.normal(0,1,10)
    ans[:,i] = ans[:,i-1] + (np.sqrt(dt))*z

for i in range(10):
    plt.plot(t,ans[i])

plt.xlabel("time")
plt.ylabel("W(t)")
plt.grid()
plt.show()

print(f"the expected value of W(2.5) = {np.mean(ans[:,2500])}")
print(f"the expected value of W(5.0) = {np.mean(ans[:,5000])}")
```

```python
#question 2
mu = [0.09,0.9,0.9]
sigma = [0.5,0.25,0.05]
np.random.seed(42)

n = 5000
dt = 5/5000
t = np.linspace(0,5,5001)

# plt.subplots(1,3)
for j in range(3):

    ans = np.zeros((10,5001))
    ans[:,0] = 5
    for i in range(1,5001):
        z = np.random.normal(0,1,10)
        ans[:,i] = ans[:,i-1] + mu[j]*dt +sigma[j]*(np.sqrt(dt))*z

    for i in range(10):
        plt.plot(t,ans[i])

    plt.xlabel("time")
    plt.ylabel("W(t)")
    plt.grid()
    plt.show()

    print(f"the expected value of W(2.5) = {np.mean(ans[:,2500])}")
    print(f"the expected value of W(5.0) = {np.mean(ans[:,5000])}")

```
